```python
import pandas as pd

data = {
    'Sky': ['Sunny', 'Sunny', 'Rainy', 'Sunny'],
    'AirTemp': ['Warm', 'Warm', 'Cold', 'Warm'],
    'Humidity': ['Normal', 'High', 'High', 'High'],
    'Wind': ['Strong', 'Strong', 'Strong', 'Strong'],
    'Water': ['Warm', 'Warm', 'Warm', 'Cool'],
    'Forecast': ['Same', 'Same', 'Change', 'Change'],
    'EnjoySport': ['Yes', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)
df.to_csv('EnjoySport.csv', index=False)

data = pd.read_csv('EnjoySport.csv')
print("Training Dataset:")
print(data)

X = data.iloc[:, :-1].values
Y = data.iloc[:, -1].values

n = X.shape[1]
domains = [data.iloc[:, i].unique() for i in range(n)]

S = {tuple([None] * n)}
G = {tuple(['?'] * n)}

def covers(h, x):
    return all(a == '?' or a == b for a, b in zip(h, x))

def more_general(h1, h2):
    return all(a == '?' or a == b for a, b in zip(h1, h2))

for x, y in zip(X, Y):
    if str(y).lower() in ['yes', 'positive', '1']:
        G = {g for g in G if covers(g, x)}
        new_S = set()
        for s in S:
            if all(v is None for v in s):
                new_S.add(tuple(x))
            else:
                h = list(s)
                for i in range(n):
                    if h[i] != x[i]:
                        h[i] = '?'
                new_S.add(tuple(h))
        S = {
            s for s in new_S
            if any(more_general(g, s) for g in G)
        }
    else:
        S = {s for s in S if not covers(s, x)}
        new_G = set()
        for g in G:
            if covers(g, x):
                for i in range(n):
                    if g[i] == '?':
                        for value in domains[i]:
                            if value != x[i]:
                                h = list(g)
                                h[i] = value
                                new_G.add(tuple(h))
            else:
                new_G.add(g)
        G = {
            g for g in new_G
            if any(more_general(g, s) for s in S)
        }

    print("\nTraining Example:", x, y)
    print("Specific Boundary S:", S)
    print("General Boundary G:", G)

print("\nFinal Specific Boundary S:")
for s in S:
    print(s)

print("\nFinal General Boundary G:")
for g in G:
    print(g)
```

```python
# Decision Tree using ID3 Algorithm

import pandas as pd
import math
import numpy as np

# Step 1: Create Dataset
data = {
    'Outlook': ['Sunny', 'Sunny', 'Overcast', 'Rain', 'Rain',
                'Rain', 'Overcast', 'Sunny', 'Sunny', 'Rain',
                'Sunny', 'Overcast', 'Overcast', 'Rain'],
    'Temperature': ['Hot', 'Hot', 'Hot', 'Mild', 'Cool',
                    'Cool', 'Cool', 'Mild', 'Cool', 'Mild',
                    'Mild', 'Mild', 'Hot', 'Mild'],
    'Humidity': ['High', 'High', 'High', 'High', 'Normal',
                 'Normal', 'Normal', 'High', 'Normal', 'Normal',
                 'Normal', 'High', 'Normal', 'High'],
    'Wind': ['Weak', 'Strong', 'Weak', 'Weak', 'Weak',
             'Strong', 'Strong', 'Weak', 'Weak', 'Weak',
             'Strong', 'Strong', 'Weak', 'Strong'],
    'PlayTennis': ['No', 'No', 'Yes', 'Yes', 'Yes',
                   'No', 'Yes', 'No', 'Yes', 'Yes',
                   'Yes', 'Yes', 'Yes', 'No']
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

# Step 2: Calculate Entropy
def entropy(target):
    values, counts = np.unique(target, return_counts=True)
    total = len(target)
    return -sum((count/total) * math.log2(count/total) for count in counts)

# Step 3: Calculate Information Gain
def information_gain(df, attribute, target):
    total_entropy = entropy(df[target])
    values, counts = np.unique(df[attribute], return_counts=True)

    weighted_entropy = 0
    for value, count in zip(values, counts):
        subset = df[df[attribute] == value]
        weighted_entropy += (count/len(df)) * entropy(subset[target])

    return total_entropy - weighted_entropy

# Step 4: Build ID3 Decision Tree
def id3(df, target, attributes):
    if len(np.unique(df[target])) == 1:
        return df[target].iloc[0]

    if len(attributes) == 0:
        return df[target].mode()[0]

    gains = [information_gain(df, attr, target) for attr in attributes]
    best_attribute = attributes[np.argmax(gains)]

    tree = {best_attribute: {}}
    for value in np.unique(df[best_attribute]):
        subset = df[df[best_attribute] == value]

        if subset.empty:
            tree[best_attribute][value] = df[target].mode()[0]
        else:
            remaining = [a for a in attributes if a != best_attribute]
            tree[best_attribute][value] = id3(subset, target, remaining)

    return tree

# Step 5: Generate Decision Tree
attributes = ['Outlook', 'Temperature', 'Humidity', 'Wind']
decision_tree = id3(df, 'PlayTennis', attributes)

print("\nDecision Tree:")
print(decision_tree)

# Step 6: Classify a New Sample
def predict(tree, sample):
    if not isinstance(tree, dict):
        return tree

    attribute = next(iter(tree))
    value = sample[attribute]

    if value not in tree[attribute]:
        return "Unknown"

    return predict(tree[attribute][value], sample)

new_sample = {
    'Outlook': 'Sunny',
    'Temperature': 'Cool',
    'Humidity': 'High',
    'Wind': 'Strong'
}

result = predict(decision_tree, new_sample)

print("\nNew Sample:")
print(new_sample)
print("Predicted Class:", result)
```
