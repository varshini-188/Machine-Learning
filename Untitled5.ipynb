```python
# Artificial Neural Network using Backpropagation Algorithm

import numpy as np

# Step 1: Create Dataset (XOR)
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([
    [0],
    [1],
    [1],
    [0]
])

# Step 2: Initialize Weights and Biases
np.random.seed(42)

input_neurons = 2
hidden_neurons = 4
output_neurons = 1

learning_rate = 0.5
epochs = 10000

W1 = np.random.randn(input_neurons, hidden_neurons)
b1 = np.zeros((1, hidden_neurons))

W2 = np.random.randn(hidden_neurons, output_neurons)
b2 = np.zeros((1, output_neurons))

# Step 3: Activation Function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Step 4: Train ANN using Backpropagation
for epoch in range(epochs):
    # Forward Propagation
    hidden_input = np.dot(X, W1) + b1
    hidden_output = sigmoid(hidden_input)

    final_input = np.dot(hidden_output, W2) + b2
    predicted_output = sigmoid(final_input)

    # Calculate Error
    error = y - predicted_output

    # Backpropagation
    d_output = error * predicted_output * (1 - predicted_output)
    d_hidden = np.dot(d_output, W2.T) * hidden_output * (1 - hidden_output)

    # Update Weights and Biases
    W2 += learning_rate * np.dot(hidden_output.T, d_output)
    b2 += learning_rate * np.sum(d_output, axis=0, keepdims=True)

    W1 += learning_rate * np.dot(X.T, d_hidden)
    b1 += learning_rate * np.sum(d_hidden, axis=0, keepdims=True)

# Step 5: Test the ANN
print("ANN Predictions:")

for i in range(len(X)):
    hidden = sigmoid(np.dot(X[i:i+1], W1) + b1)
    output = sigmoid(np.dot(hidden, W2) + b2)
    predicted_class = 1 if output[0][0] >= 0.5 else 0
    print("Input:", X[i], "Predicted:", predicted_class, "Actual:", y[i][0])

# Step 6: Calculate Accuracy
hidden_output = sigmoid(np.dot(X, W1) + b1)
final_output = sigmoid(np.dot(hidden_output, W2) + b2)
predicted_classes = (final_output >= 0.5).astype(int)
accuracy = np.mean(predicted_classes == y) * 100

print("\nAccuracy:", accuracy, "%")
```
