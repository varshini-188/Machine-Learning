```markdown
# Credit Score Classification

This notebook demonstrates a complete workflow for classifying customer credit scores into categories (**Poor**, **Standard**, **Good**) using Machine Learning.

### Workflow Steps:
1. **Data Generation**: Create a synthetic dataset representing financial profiles.
2. **Exploratory Data Analysis (EDA)**: Inspect the dataset structure.
3. **Data Preprocessing**: Handle categorical variables and scale numerical features.
4. **Model Training**: Train a Machine Learning classifier (Random Forest).
5. **Model Evaluation**: Evaluate the classifier using classification metrics and a confusion matrix.
```

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# Set random seed for reproducibility
np.random.seed(42)

# 1. Generate Synthetic Credit Data
n_samples = 1000

data = {
    'Age': np.random.randint(18, 70, size=n_samples),
    'Annual_Income': np.random.normal(55000, 20000, size=n_samples).clip(10000, 150000),
    'Num_Bank_Accounts': np.random.randint(1, 10, size=n_samples),
    'Num_Credit_Cards': np.random.randint(1, 12, size=n_samples),
    'Interest_Rate': np.random.uniform(5, 35, size=n_samples),
    'Delay_from_due_date': np.random.randint(0, 30, size=n_samples),
    'Num_of_Delayed_Payment': np.random.randint(0, 25, size=n_samples),
    'Outstanding_Debt': np.random.uniform(100, 5000, size=n_samples),
    'Credit_History_Age_Months': np.random.randint(6, 360, size=n_samples),
    'Monthly_Balance': np.random.uniform(100, 2000, size=n_samples)
}

df = pd.DataFrame(data)

# Define a rule-based engine to assign synthetic Credit Scores
def assign_credit_score(row):
    score = 0
    # Income & Debt ratio heuristic
    if row['Annual_Income'] > 80000: score += 3
    elif row['Annual_Income'] > 40000: score += 2
    else: score += 1

    if row['Outstanding_Debt'] < 1000: score += 3
    elif row['Outstanding_Debt'] < 2500: score += 2
    else: score += 0

    # Delays hurt score significantly
    if row['Delay_from_due_date'] < 5: score += 3
    elif row['Delay_from_due_date'] < 15: score += 1
    else: score -= 2

    if row['Credit_History_Age_Months'] > 180: score += 2
    elif row['Credit_History_Age_Months'] > 60: score += 1

    if score >= 8: return 'Good'
    elif score >= 4: return 'Standard'
    else: return 'Poor'

df['Credit_Score'] = df.apply(assign_credit_score, axis=1)

# Display the first 5 rows
display(df.head())
````

```markdown
### 2. Exploratory Data Analysis
Let's check the distribution of our target variable `Credit_Score`.
```

````python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Credit_Score', order=['Poor', 'Standard', 'Good'], palette='viridis')
plt.title('Distribution of Credit Scores')
plt.show()
````

```markdown
### 3. Preprocessing and Feature Engineering
We need to split the data into training and test sets, encode the target label, and scale the features to optimize classification performance.
```

````python
# Separate Features (X) and Target (y)
X = df.drop(columns=['Credit_Score'])
y = df['Credit_Score']

# Encode target labels ('Poor', 'Standard', 'Good') to integers (0, 1, 2)
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42, stratify=y_encoded)

# Standardize numerical features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(f"Training set shape: {X_train.shape}")
print(f"Testing set shape: {X_test.shape}")
````

```markdown
### 4. Model Training & Evaluation
We will train a **Random Forest Classifier** and evaluate its performance with a classification report and confusion matrix.
```

````python
# Initialize and train Random Forest Classifier
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)
rf_classifier.fit(X_train_scaled, y_train)

# Predict on the test set
y_pred = rf_classifier.predict(X_test_scaled)

# Evaluate the Model
accuracy = accuracy_score(y_test, y_pred)
print(f"Overall Classification Accuracy: {accuracy * 100:.2f}%\n")

print("Classification Report:")
print(classification_report(y_test, y_pred, target_names=label_encoder.classes_))

# Plot Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', xticklabels=label_encoder.classes_, yticklabels=label_encoder.classes_, cmap='Blues')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.show()
````

```markdown
### 5. Feature Importance
Let's check which financial features had the greatest impact on determining the Credit Score class.
```

````python
importances = rf_classifier.feature_importances_
indices = np.argsort(importances)[::-1]

plt.figure(figsize=(10, 6))
sns.barplot(x=importances[indices], y=X.columns[indices], palette='mako')
plt.title('Feature Importance in Credit Score Classification')
plt.xlabel('Importance Value')
plt.ylabel('Feature Name')
plt.show()
````
