```python
# Artificial Neural Network using Backpropagation Algorithm

import numpy as np

# Step 1: Create Dataset (XOR)
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([
    [0],
    [1],
    [1],
    [0]
])

# Step 2: Initialize Weights and Biases
np.random.seed(42)

input_neurons = 2
hidden_neurons = 4
output_neurons = 1

learning_rate = 0.5
epochs = 10000

W1 = np.random.randn(input_neurons, hidden_neurons)
b1 = np.zeros((1, hidden_neurons))

W2 = np.random.randn(hidden_neurons, output_neurons)
b2 = np.zeros((1, output_neurons))

# Step 3: Activation Function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Step 4: Train ANN using Backpropagation
for epoch in range(epochs):
    # Forward Propagation
    hidden_input = np.dot(X, W1) + b1
    hidden_output = sigmoid(hidden_input)

    final_input = np.dot(hidden_output, W2) + b2
    predicted_output = sigmoid(final_input)

    # Calculate Error
    error = y - predicted_output

    # Backpropagation
    d_output = error * predicted_output * (1 - predicted_output)
    d_hidden = np.dot(d_output, W2.T) * hidden_output * (1 - hidden_output)

    # Update Weights and Biases
    W2 += learning_rate * np.dot(hidden_output.T, d_output)
    b2 += learning_rate * np.sum(d_output, axis=0, keepdims=True)

    W1 += learning_rate * np.dot(X.T, d_hidden)
    b1 += learning_rate * np.sum(d_hidden, axis=0, keepdims=True)

# Step 5: Test the ANN
print("ANN Predictions:")

for i in range(len(X)):
    hidden = sigmoid(np.dot(X[i:i+1], W1) + b1)
    output = sigmoid(np.dot(hidden, W2) + b2)
    predicted_class = 1 if output[0][0] >= 0.5 else 0
    print("Input:", X[i], "Predicted:", predicted_class, "Actual:", y[i][0])

# Step 6: Calculate Accuracy
hidden_output = sigmoid(np.dot(X, W1) + b1)
final_output = sigmoid(np.dot(hidden_output, W2) + b2)
predicted_classes = (final_output >= 0.5).astype(int)
accuracy = np.mean(predicted_classes == y) * 100

print("\nAccuracy:", accuracy, "%")
```

### Implementation of K-Nearest Neighbors (K-NN) from Scratch
In this section, we define a standard K-NN classifier utilizing Euclidean distance.

```python
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

class KNNClassifier:
    def __init__(self, k=3):
        self.k = k

    def fit(self, X, y):
        self.X_train = X
        self.y_train = y

    def predict(self, X):
        predictions = [self._predict_single(x) for x in X]
        return np.array(predictions)

    def _predict_single(self, x):
        # 1. Compute Euclidean distances between x and all examples in the training set
        distances = [np.sqrt(np.sum((x - x_train) ** 2)) for x_train in self.X_train]

        # 2. Get the indices of the k nearest neighbors
        k_indices = np.argsort(distances)[:self.k]

        # 3. Extract the labels of the k nearest neighbors
        k_nearest_labels = [self.y_train[i] for i in k_indices]

        # 4. Return the most common class label among the neighbors
        most_common = Counter(k_nearest_labels).most_common(1)
        return most_common[0][0]
```

### Generate Synthetic Dataset and Train/Evaluate the Model

```python
# Create a simple 2D synthetic dataset with 2 classes
np.random.seed(42)

# Class 0 data
class_0 = np.random.randn(20, 2) + np.array([1, 1])
y_0 = np.zeros(20, dtype=int)

# Class 1 data
class_1 = np.random.randn(20, 2) + np.array([4, 4])
y_1 = np.ones(20, dtype=int)

# Merge into a single dataset
X = np.vstack((class_0, class_1))
y = np.concatenate((y_0, y_1))

# Fit our K-NN Model
knn = KNNClassifier(k=5)
knn.fit(X, y)

# Define some test points to make predictions on
test_points = np.array([
    [1.5, 1.5],
    [3.0, 3.0],
    [4.5, 4.0]
])
predictions = knn.predict(test_points)

# Print predictions
for point, pred in zip(test_points, predictions):
    print(f"Input query point {point} -> Predicted Class: {pred}")
```

### Visualize Decision Boundary and Test Points

```python
# Plot original dataset classes
plt.figure(figsize=(10, 6))
plt.scatter(X[y == 0][:, 0], X[y == 0][:, 1], color='blue', label='Class 0 (Training)', s=80, edgecolors='k')
plt.scatter(X[y == 1][:, 0], X[y == 1][:, 1], color='red', label='Class 1 (Training)', s=80, edgecolors='k')

# Plot the test query points and their predictions
colors = {0: 'lightblue', 1: 'salmon'}
for i, (point, pred) in enumerate(zip(test_points, predictions)):
    plt.scatter(point[0], point[1], color=colors[pred], marker='*', s=250, edgecolors='black', linewidth=1.5,
                label=f'Test Point {i+1} (Pred: {pred})')

plt.title('K-Nearest Neighbors (K-NN) Visualization (k=5)')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
