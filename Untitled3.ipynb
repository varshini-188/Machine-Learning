```python
import pandas as pd

data = {
    'Sky': ['Sunny', 'Sunny', 'Rainy', 'Sunny'],
    'AirTemp': ['Warm', 'Warm', 'Cold', 'Warm'],
    'Humidity': ['Normal', 'High', 'High', 'High'],
    'Wind': ['Strong', 'Strong', 'Strong', 'Strong'],
    'Water': ['Warm', 'Warm', 'Warm', 'Cool'],
    'Forecast': ['Same', 'Same', 'Change', 'Change'],
    'EnjoySport': ['Yes', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)
df.to_csv('EnjoySport.csv', index=False)

data = pd.read_csv('EnjoySport.csv')
print("Training Dataset:")
print(data)

X = data.iloc[:, :-1].values
Y = data.iloc[:, -1].values

n = X.shape[1]
domains = [data.iloc[:, i].unique() for i in range(n)]

S = {tuple([None] * n)}
G = {tuple(['?'] * n)}

def covers(h, x):
    return all(a == '?' or a == b for a, b in zip(h, x))

def more_general(h1, h2):
    return all(a == '?' or a == b for a, b in zip(h1, h2))

for x, y in zip(X, Y):
    if str(y).lower() in ['yes', 'positive', '1']:
        G = {g for g in G if covers(g, x)}
        new_S = set()
        for s in S:
            if all(v is None for v in s):
                new_S.add(tuple(x))
            else:
                h = list(s)
                for i in range(n):
                    if h[i] != x[i]:
                        h[i] = '?'
                new_S.add(tuple(h))
        S = {
            s for s in new_S
            if any(more_general(g, s) for g in G)
        }
    else:
        S = {s for s in S if not covers(s, x)}
        new_G = set()
        for g in G:
            if covers(g, x):
                for i in range(n):
                    if g[i] == '?':
                        for value in domains[i]:
                            if value != x[i]:
                                h = list(g)
                                h[i] = value
                                new_G.add(tuple(h))
            else:
                new_G.add(g)
        G = {
            g for g in new_G
            if any(more_general(g, s) for s in S)
        }

    print("\nTraining Example:", x, y)
    print("Specific Boundary S:", S)
    print("General Boundary G:", G)

print("\nFinal Specific Boundary S:")
for s in S:
    print(s)

print("\nFinal General Boundary G:")
for g in G:
    print(g)
```
