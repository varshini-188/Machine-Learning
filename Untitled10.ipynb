```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 1. Generate synthetic data from two Gaussians
np.random.seed(42)

# True parameters
mu1_true, sigma1_true = 2.0, 1.0
mu2_true, sigma2_true = 7.0, 1.5
weight_true = 0.6  # fraction of data from cluster 1

N = 400
# Determine how many points come from each cluster
n1 = int(N * weight_true)
n2 = N - n1

data1 = np.random.normal(mu1_true, sigma1_true, n1)
data2 = np.random.normal(mu2_true, sigma2_true, n2)
X = np.concatenate([data1, data2])

# Plot the data
plt.figure(figsize=(10, 4))
plt.hist(X, bins=30, density=True, alpha=0.6, color='g', edgecolor='black')
plt.title("Synthetic Dataset Histogram")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

```python
def em_algorithm(X, max_iter=100, tol=1e-6):
    # Initialize parameters
    # Pick random points from dataset as initial means
    mu1 = np.random.choice(X)
    mu2 = np.random.choice(X)
    # Initialize variances to 1
    var1 = 1.0
    var2 = 1.0
    # Initialize mixing coefficient
    pi1 = 0.5

    print(f"Initial Guesses: mu1={mu1:.2f}, var1={var1:.2f}, mu2={mu2:.2f}, var2={var2:.2f}, pi1={pi1:.2f}\n")

    for i in range(max_iter):
        prev_mu1, prev_mu2 = mu1, mu2

        # --- E-Step: Compute responsibilities ---
        # Likelihood of each data point under cluster 1 and cluster 2
        likelihood1 = norm.pdf(X, mu1, np.sqrt(var1))
        likelihood2 = norm.pdf(X, mu2, np.sqrt(var2))

        # Responsibilities (posterior probabilities)
        denom = pi1 * likelihood1 + (1 - pi1) * likelihood2
        gamma1 = (pi1 * likelihood1) / denom
        gamma2 = ((1 - pi1) * likelihood2) / denom

        # --- M-Step: Update parameters ---
        N1 = np.sum(gamma1)
        N2 = np.sum(gamma2)

        # Update weights
        pi1 = N1 / len(X)

        # Update means
        mu1 = np.sum(gamma1 * X) / N1
        mu2 = np.sum(gamma2 * X) / N2

        # Update variances
        var1 = np.sum(gamma1 * (X - mu1)**2) / N1
        var2 = np.sum(gamma2 * (X - mu2)**2) / N2

        # Check convergence
        if np.abs(mu1 - prev_mu1) < tol and np.abs(mu2 - prev_mu2) < tol:
            print(f"Converged in {i+1} iterations.")
            break

    return mu1, np.sqrt(var1), mu2, np.sqrt(var2), pi1

# Run the EM algorithm
mu1_est, std1_est, mu2_est, std2_est, pi1_est = em_algorithm(X)

print(f"\nEstimated Parameters:")
print(f"Cluster 1: mean = {mu1_est:.3f} (True: {mu1_true}), std = {std1_est:.3f} (True: {sigma1_true})")
print(f"Cluster 2: mean = {mu2_est:.3f} (True: {mu2_true}), std = {std2_est:.3f} (True: {sigma2_true})")
print(f"Weight Cluster 1: {pi1_est:.3f} (True: {weight_true})")
```

```python
# Visualize the estimated model vs the data
x_axis = np.linspace(X.min() - 2, X.max() + 2, 1000)

# Estimated PDFs
curve1 = pi1_est * norm.pdf(x_axis, mu1_est, std1_est)
curve2 = (1 - pi1_est) * norm.pdf(x_axis, mu2_est, std2_est)
mixture = curve1 + curve2

plt.figure(figsize=(10, 5))
plt.hist(X, bins=30, density=True, alpha=0.5, color='gray', label='Data Density')
plt.plot(x_axis, curve1, 'r--', label='Estimated G1')
plt.plot(x_axis, curve2, 'b--', label='Estimated G2')
plt.plot(x_axis, mixture, 'k-', linewidth=2, label='Estimated Mixture')
plt.title("EM Algorithm Final Estimation")
plt.xlabel("X value")
plt.ylabel("Density")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
