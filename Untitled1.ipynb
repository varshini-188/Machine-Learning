### Sample Dataset Creation
Since we don't have an external dataset uploaded yet, let's create a synthetic DataFrame representing common features that influence car prices (e.g., Year, Mileage, Engine Size, Fuel Type, Brand, and Price).

```python
# Generate synthetic dataset for demonstration
np.random.seed(42)
n_samples = 200

data = {
    'Brand': np.random.choice(['Toyota', 'Honda', 'Ford', 'BMW', 'Audi'], n_samples),
    'Year': np.random.randint(2010, 2023, n_samples),
    'Mileage': np.random.randint(5000, 150000, n_samples),
    'Engine_Size': np.round(np.random.uniform(1.0, 4.0, n_samples), 1),
    'Fuel_Type': np.random.choice(['Petrol', 'Diesel', 'Hybrid'], n_samples),
}

df = pd.DataFrame(data)

# Define a relationship for Price based on features with some noise
df['Price'] = (
    (df['Year'] - 2010) * 1500
    - (df['Mileage'] * 0.05)
    + (df['Engine_Size'] * 4000)
    + np.random.normal(15000, 2000, n_samples)
)
df['Price'] = df['Price'].clip(lower=2000) # Ensure no negative prices

display(df.head())
```
